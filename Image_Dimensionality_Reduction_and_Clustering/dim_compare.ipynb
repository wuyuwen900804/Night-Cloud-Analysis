```python
import pandas as pd
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
df = pd.read_csv("./dimension_records.csv")
df = df.loc[:,:]
```

```python
n_components_range = list(range(2, 10))
highlight_k = 6
highlight_index = n_components_range.index(highlight_k)

fig = make_subplots(
    rows=1, cols=2,
    horizontal_spacing=0.15,
    vertical_spacing=0
)

# --- Subplot Data ---
traces = [
    {'y': df['AvgLoss'], 'name': 'RMSE', 'row': 1, 'col': 1, 'y_label': '<b>RMSE value</b>', 'label': 'a)'},
    {'y': df['AvgR2'], 'name': 'R²', 'row': 1, 'col': 2, 'y_label': '<b>R² value</b>', 'label': 'b)'},
]

for t in traces:
    fig.add_trace(
        go.Scatter(
            x=n_components_range, 
            y=t['y'], 
            mode='lines+markers', 
            name=t['name'],
            marker=dict(color='#999999', size=12),
            line=dict(color='#999999', width=4)
        ),
        row=t['row'], col=t['col']
    )

    fig.add_trace(
        go.Scatter(
            x=[highlight_k], 
            y=[t['y'][highlight_index]], 
            mode='markers',
            marker=dict(color='#3AA396', size=24, symbol='circle-open', line=dict(width=4)),
            name=f"Selected k={highlight_k}"
        ),
        row=t['row'], col=t['col']
    )

    fig.update_yaxes(title_text=t['y_label'], row=t['row'], col=t['col'], title_font=dict(size=36, family="Arial", weight="bold"))
    
    fig.add_annotation(
        text=f"<b>{t['label']}</b>",
        xref=f"x{t['col'] if t['col'] > 1 else ''} domain",
        yref=f"y{t['col'] if t['col'] > 1 else ''} domain",
        x=0.03, y=0.97,
        showarrow=False,
        font=dict(size=32, family="Arial", weight="bold"),
        xanchor='left',
        yanchor='top'
    )

fig.update_yaxes(tickformat=".3f", row=1, col=1)
fig.update_yaxes(tickformat=".3f", row=1, col=2)

fig.update_layout(
    height=450,
    width=1200,
    showlegend=False,
    template='simple_white',
    margin=dict(t=10, b=110, l=10, r=10),
    font=dict(family="Arial",size=28,weight="bold")
)

fig.update_xaxes(linewidth=2.5, linecolor='black')
fig.update_yaxes(linewidth=2.5, linecolor='black')

fig.add_annotation(
    text="<b>Dimension</b>",
    xref="paper", yref="paper",
    x=0.5, y=-0.18,
    showarrow=False,
    font=dict(size=36, family="Arial", weight="bold"),
    xanchor='center',
    yanchor='top'
)

fig.show()
```
