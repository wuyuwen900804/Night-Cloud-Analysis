```python
n_components_range = list(range(2, 21))
highlight_k = 3
highlight_index = n_components_range.index(highlight_k)

fig = make_subplots(
    rows=1, cols=4,
    horizontal_spacing=0.07,
    vertical_spacing=0
)

# --- Subplot Data ---
traces = [
    {'y': sse, 'name': 'SSE', 'row': 1, 'col': 1, 'y_label': '<b>SSE</b>', 'label': 'a)'},
    {'y': davies_bouldin, 'name': 'Davies-Bouldin', 'row': 1, 'col': 2, 'y_label': '<b>DB Index</b>', 'label': 'b)'},
    {'y': calinski_harabasz, 'name': 'Calinski-Harabasz', 'row': 1, 'col': 3, 'y_label': '<b>CH Index</b>', 'label': 'c)'},
    {'y': silhouette, 'name': 'Silhouette Score', 'row': 1, 'col': 4, 'y_label': '<b>Silhouette Score</b>', 'label': 'd)'}
]

for t in traces:
    fig.add_trace(
        go.Scatter(
            x=n_components_range, 
            y=t['y'], 
            mode='lines+markers', 
            name=t['name'],
            marker=dict(color='#999999', size=12),
            line=dict(color='#999999', width=4)
        ),
        row=t['row'], col=t['col']
    )

    fig.add_trace(
        go.Scatter(
            x=[highlight_k], 
            y=[t['y'][highlight_index]], 
            mode='markers',
            marker=dict(color='#3AA396', size=24, symbol='circle-open', line=dict(width=4)),
            name=f"Selected k={highlight_k}"
        ),
        row=t['row'], col=t['col']
    )

    fig.update_yaxes(title_text=t['y_label'], row=t['row'], col=t['col'], title_font=dict(size=36, family="Arial", weight="bold"))
    
    fig.add_annotation(
        text=f"<b>{t['label']}</b>",
        xref=f"x{t['col'] if t['col'] > 1 else ''} domain",
        yref=f"y{t['col'] if t['col'] > 1 else ''} domain",
        x=0.03, y=0.97,
        showarrow=False,
        font=dict(size=32, family="Arial", weight="bold"),
        xanchor='left',
        yanchor='top'
    )

fig.update_yaxes(tickformat=".1f", row=1, col=2)
fig.update_yaxes(range=[10, 1.35*10e5], row=1, col=1)
fig.update_yaxes(range=[-0.05, 11], row=1, col=2)
fig.update_yaxes(range=[4*10e2, 17*10e2], row=1, col=3)
fig.update_yaxes(range=[-0.25, 0.85], row=1, col=4)


fig.update_layout(
    height=450,
    width=2200,
    showlegend=False,
    template='simple_white',
    margin=dict(t=10, b=110, l=10, r=10),
    font=dict(family="Arial",size=28,weight="bold")
)

fig.update_xaxes(linewidth=2.5, linecolor='black')
fig.update_yaxes(linewidth=2.5, linecolor='black')

fig.add_annotation(
    text="<b>Number of clusters</b>",
    xref="paper", yref="paper",
    x=0.5, y=-0.18,
    showarrow=False,
    font=dict(size=36, family="Arial", weight="bold"),
    xanchor='center',
    yanchor='top'
)

fig.show()
```
