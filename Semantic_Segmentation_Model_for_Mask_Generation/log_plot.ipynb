```python
import torch
import random
import numpy as np
import pandas as pd
import plotly.graph_objs as go
```

```python
def set_seed(seed):
    np.random.seed(seed)
    random.seed(seed)
    torch.manual_seed(seed)
    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False
    if torch.cuda.is_available():
        torch.cuda.manual_seed_all(seed)
        torch.cuda.manual_seed(seed)
set_seed(0)
```

```python
def plot_loss(history):
    train_loss_trace = go.Scatter(
        x=list(range(len(history['train_loss']))),
        y=history['train_loss'],
        mode='lines',
        name=' Training Loss',
        line=dict(color='#2a9d8f',width=10)
        # line=dict(color='#89C65C',width=10)
    )
    val_loss_trace = go.Scatter(
        x=list(range(len(history['val_loss']))),
        y=history['val_loss'],
        mode='lines',
        name=' Validation Loss',
        line=dict(color='#e76f51',width=10)
        # line=dict(color='#39BCBC',width=10)
    )
    data = [train_loss_trace, val_loss_trace]
    fig = go.Figure(data=data)
    fig.update_layout(
        title=None,
        yaxis_range=[-0.1, 7.5],
        xaxis_title='Epochs', yaxis_title='Loss',
        xaxis=dict(tickmode='linear', dtick=20),
        legend=dict(x=0.98, y=0.96, xanchor='right', yanchor='top', font=dict(size=56), itemwidth=100),
        margin=dict(l=200, r=10, t=15, b=180),
        height=1000, width=1200,
        font=dict(size=56, family="Arial", weight="bold"),
        annotations=[
            dict(
                text="a)",
                x=0.03, y=0.97,
                xref="paper", yref="paper",
                showarrow=False,
                font=dict(size=80, family="Arial", weight="bold")
            )
        ],
        template='simple_white'
    )
    fig.update_xaxes(linewidth=5, linecolor='black', tickfont=dict(size=56), title_standoff=30)
    fig.update_yaxes(linewidth=5, linecolor='black', tickfont=dict(size=56), title_standoff=30)
    fig.show()
    
def plot_acc(history):
    train_acc_trace = go.Scatter(
        x=list(range(len(history['train_acc']))),
        y=history['train_acc'],
        mode='lines',
        name=' Training Accuracy',
        line=dict(color='#2a9d8f',width=10)
        # line=dict(color='#89C65C',width=10)
    )
    val_acc_trace = go.Scatter(
        x=list(range(len(history['val_acc']))),
        y=history['val_acc'],
        mode='lines',
        name=' Validation Accuracy',
        line=dict(color='#e76f51',width=10)
        # line=dict(color='#39BCBC',width=10)
    )
    data = [train_acc_trace, val_acc_trace]
    fig = go.Figure(data=data)
    fig.update_layout(
        title=None,
        xaxis_title='Epochs', yaxis_title='Acc',
        xaxis=dict(tickmode='linear', dtick=20),
        legend=dict(x=0.98, y=0.04, xanchor='right', yanchor='bottom', font=dict(size=56), itemwidth=100),
        margin=dict(l=200, r=10, t=15, b=180),
        height=1000, width=1200,
        font=dict(size=56, family="Arial", weight="bold"),
        annotations=[
            dict(
                text="b)",
                x=0.03, y=0.97,
                xref="paper", yref="paper",
                showarrow=False,
                font=dict(size=80, family="Arial", weight="bold")
            )
        ],
        template='simple_white'
    )
    fig.update_xaxes(linewidth=5, linecolor='black', tickfont=dict(size=56), title_standoff=30)
    fig.update_yaxes(linewidth=5, linecolor='black', tickfont=dict(size=56), title_standoff=30)
    fig.show()

def plot_score(history):
    train_wiou_trace = go.Scatter(
        x=list(range(len(history['train_wiou']))),
        y=history['train_wiou'],
        mode='lines',
        name=' Training wIoU',
        line=dict(color='#2a9d8f',width=10)
        # line=dict(color='#89C65C',width=10)
    )
    val_wiou_trace = go.Scatter(
        x=list(range(len(history['val_wiou']))),
        y=history['val_wiou'],
        mode='lines',
        name=' Validation wIoU',
        line=dict(color='#e76f51',width=10)
        # line=dict(color='#39BCBC',width=10)
    )
    data = [train_wiou_trace, val_wiou_trace]
    fig = go.Figure(data=data)
    fig.update_layout(
        title=None,
        xaxis_title='Epochs', yaxis_title='wIoU',
        xaxis=dict(tickmode='linear', dtick=20),
        legend=dict(x=0.98, y=0.04, xanchor='right', yanchor='bottom', font=dict(size=56), itemwidth=100),
        margin=dict(l=200, r=10, t=15, b=180),
        height=1000, width=1200,
        font=dict(size=56, family="Arial", weight="bold"),
        annotations=[
            dict(
                text="c)",
                x=0.03, y=0.97,
                xref="paper", yref="paper",
                showarrow=False,
                font=dict(size=80, family="Arial", weight="bold")
            )
        ],
        template='simple_white'
    )
    fig.update_xaxes(linewidth=5, linecolor='black', tickfont=dict(size=56), title_standoff=30)
    fig.update_yaxes(linewidth=5, linecolor='black', tickfont=dict(size=56), title_standoff=30)
    fig.show()
```

```python
history = pd.read_csv('training_history_target.csv')
```

```python
plot_loss(history)
```

```python
plot_acc(history)
```

```python
plot_score(history)
```
