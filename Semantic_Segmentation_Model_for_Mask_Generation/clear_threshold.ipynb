```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

img = cv2.imread("./Main/IMG_0356.jpg")
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
output = np.zeros((gray.shape[0], gray.shape[1], 3), dtype=np.uint8)

# if gray <= 72, set to black; else set to yellow
output[gray <= 72] = [0, 0, 0]
output[gray > 72] = [255, 255, 0]

plt.figure(figsize=(12,5))

plt.subplot(1,3,1)
plt.title("Original")
plt.imshow(img_rgb)
plt.axis("off")

plt.subplot(1,3,2)
plt.title("Gray")
plt.imshow(gray, cmap="gray")
plt.axis("off")

plt.subplot(1,3,3)
plt.title("<=72=Black, >72=Yellow")
plt.imshow(output)
plt.axis("off")

plt.show()
```
