```python
import cv2
import numpy as np

img_path = './Main_mask/IMG_0001_mask.png'
img = cv2.imread(img_path)

old_color = np.array([225, 225, 225])
new_color = np.array([0, 0, 0])

mask = np.all(img == old_color, axis=-1)
img[mask] = new_color

cv2.imwrite(img_path, img)

print("Color replacement completed.")
```
