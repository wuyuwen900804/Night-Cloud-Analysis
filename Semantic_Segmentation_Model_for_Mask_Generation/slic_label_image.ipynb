```python
import cv2
import numpy as np
from skimage.segmentation import slic, mark_boundaries
from skimage.measure import regionprops
import matplotlib.pyplot as plt
```

```python
num = '0001' # specify the image number
n_segments = 150  # assign the number of superpixels
compactness = 10  # control the compactness of superpixels
```

```python
image_path = f'Main/IMG_{num}.jpg'
image = cv2.imread(image_path)
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

bright_factor = 1  # add a brightness factor (1 means no change)
image = np.clip(image * bright_factor, 0, 255).astype(np.uint8)

# SLIC superpixel
segments = slic(image, n_segments=n_segments, compactness=compactness, start_label=1)

labeled_image = image.copy()

props = regionprops(segments)
for prop in props:
    y, x = prop.centroid
    label = prop.label
    cv2.putText(
        labeled_image, 
        str(label), 
        (int(x)-40, int(y)), 
        cv2.FONT_HERSHEY_SIMPLEX, 
        2,  # font size
        (0, 0, 0),  # font color (black)
        6,  # font thickness
        cv2.LINE_AA
    )

marked_image = mark_boundaries(labeled_image, segments)

plt.figure(figsize=(15, 10))
plt.subplot(1, 3, 1)
plt.title("Original Image")
plt.imshow(image)
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Labeled Superpixels")
plt.imshow(labeled_image)
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Superpixel Segmentation with Labels")
plt.imshow(marked_image)
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# [255, 75, 75] Cirrus
# [255, 150, 0] Cirrostratus
# [255, 210, 75] Stratus
# [175, 225, 130] Stratocumulus
# [130, 175, 225] Cumulus
# [175, 150, 255] Cirrocumulus
# [225, 225, 225] Thin
# [150, 150, 150] Nimbus
```

```python
output_path = f'Main_mask/IMG_{num}_mask.png'

image = cv2.imread(image_path)
if image is None:
    raise FileNotFoundError(f"Unable to read image. Please check the path: {image_path}")
image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image = np.clip(image * bright_factor, 0, 255).astype(np.uint8)

# SLIC superpixel
segments = slic(image, n_segments=n_segments, compactness=compactness, start_label=1)

# choose the target labels for coloring (example: 1, 2, 3, 4, 5, 6, 7, 8)
# target_labels = []
target_labels = [x for x in range(1, 160) if x not in {}]

# add a new img to show the colored superpixels
colored_image = np.zeros_like(image)  # initialize a blank image for coloring

# assign colors
color_real = image.copy()
mask = np.isin(segments, target_labels)
colored_image[mask] = color_real[mask] = [175, 150, 255]

# second color
target_labels2 = []
mask2 = np.isin(segments, target_labels2)
colored_image[mask2] = color_real[mask2] = [255, 210, 75]

# third color
target_labels3 = []
mask3 = np.isin(segments, target_labels3)
colored_image[mask3] = color_real[mask3] = [0, 0, 0]

# fourth color
target_labels4 = []
mask4 = np.isin(segments, target_labels4)
colored_image[mask4] = color_real[mask4] = [175, 150, 255]

marked_image = mark_boundaries(color_real, segments)

plt.figure(figsize=(15, 10))
plt.subplot(1, 3, 1)
plt.title("Original Image")
plt.imshow(image)
plt.axis("off")

plt.subplot(1, 3, 2)
plt.title("Colored Superpixel") 
plt.imshow(colored_image)
plt.axis("off")

plt.subplot(1, 3, 3)
plt.title("Superpixel Segmentation with Boundaries")
plt.imshow(marked_image)
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
output_path = f'Main_mask/IMG_{num}_mask.png'
colored_image_bgr = cv2.cvtColor(colored_image, cv2.COLOR_RGB2BGR)
cv2.imwrite(output_path, colored_image_bgr)
print(f"The image is saved to: {output_path}")
```
