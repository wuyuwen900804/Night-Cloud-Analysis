```python
import pandas as pd
import plotly.graph_objects as go
```

```python
df = pd.read_csv("cluster_attribute.csv")
```

```python
# cloud_types = ['Cirrus', 'Cirrostratus', 'Stratus', 'Stratocumulus',
#                'Cumulus', 'Cirrocumulus', 'Nimbus', 'Clear']
cloud_types = [
    'Cirrus_weighted', 'Cirrostratus_weighted', 'Stratus_weighted', 'Stratocumulus_weighted',
    'Cumulus_weighted', 'Cirrocumulus_weighted', 'Nimbus_weighted', 'Clear_weighted'
]

colors = [
    '#FF4B4B','#FF9600','#FFD24B','#AFE182',
    '#82AFE1','#AF96FF','#969696','#000000'
]

monthly_sum = df.groupby('Month')[cloud_types].sum()
month_order = [3, 2, 1, 12, 11, 10]

existing_months_in_order = [m for m in month_order if m in monthly_sum.index]
monthly_sum = monthly_sum.reindex(existing_months_in_order)

fig = go.Figure()

# add a bar for each cloud type
for i, cloud_type in enumerate(cloud_types):
    fig.add_trace(go.Bar(
        y=monthly_sum.index, x=monthly_sum[cloud_type],
        name=cloud_type, marker_color=colors[i],
        orientation='h', showlegend=False
    ))

for cloud_type, color in zip(reversed(cloud_types), reversed(colors)):
    fig.add_trace(go.Scatter(
        x=[None], y=[None], mode='markers',
        marker=dict(size=12, color=color, symbol="square"),
        name=cloud_type.replace('_weighted', ''), showlegend=True, hoverinfo='skip'
    ))

fig.update_layout(
    barmode='stack',
    barnorm='percent',
    title_text='Monthly (Weighted) Cloud Type Distribution',
    title_x=0.5, title_y=0.95,
    yaxis_title='Month',
    xaxis_title='Percentage',
    legend_title='Type',
    font=dict(
        size=16
    ),
    yaxis=dict(
        type='category',
        categoryorder='array',
        categoryarray=monthly_sum.index
    ),
    width=1200, height=400,
    margin=dict(l=50, r=30, t=80, b=50),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        x=1, y=1.02,
        xanchor="right"
    )
)

fig.show()
```

```python
monthly_sum = df.groupby('Month')[cloud_types].sum()
month_order = [10, 11, 12, 1, 2, 3, 4]
existing_months_in_order = [m for m in month_order if m in monthly_sum.index]
monthly_sum = monthly_sum.reindex(existing_months_in_order)

monthly_percent = monthly_sum.div(monthly_sum.sum(axis=1), axis=0) * 100
monthly_percent = monthly_percent.round(1)
monthly_percent.columns = [ct.replace('_weighted','') for ct in monthly_percent.columns]
monthly_percent = monthly_percent.astype(str) + '%'
monthly_percent
```

```python
# Define label order
labels = [1, 2, 3]

# Calculate the count of each label for each month
monthly_label_counts = df.groupby(['Month', 'label']).size().unstack(fill_value=0)

# Define month order
month_order = [10, 11, 12, 1, 2, 3]

# Reindex to sort months and ensure all months are present
existing_months_in_order = [m for m in month_order if m in monthly_label_counts.index]
monthly_label_counts = monthly_label_counts.reindex(existing_months_in_order)

fig = go.Figure()

# Add a bar for each label
for label in labels:
    if label in monthly_label_counts.columns:
        fig.add_trace(go.Bar(
            x=monthly_label_counts.index,
            y=monthly_label_counts[label],
            name=str(label)
        ))

fig.update_layout(
    barmode='stack',
    barnorm='percent',
    title_text='Monthly Cluster Distribution',
    title_x=0.5, title_y=0.95,
    xaxis_title='Month',
    yaxis_title='Percentage',
    legend_title='Cluster',
    font=dict(
        size=16
    ),
    xaxis=dict(
        type='category',
        categoryorder='array',
        categoryarray=monthly_label_counts.index
    ),
    width=500, height=400,
    margin=dict(l=50, r=20, t=80, b=50),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        y=1.02,
        xanchor="right",
        x=0.83
    )
)

fig.show()
```

```python
df = pd.read_csv("cluster_attribute_case.csv")
```

```python
# cloud_types = ['Cirrus', 'Cirrostratus', 'Stratus', 'Stratocumulus',
#                'Cumulus', 'Cirrocumulus', 'Nimbus', 'Clear']
cloud_types = [
    'Cirrus_weighted', 'Cirrostratus_weighted', 'Stratus_weighted', 'Stratocumulus_weighted',
    'Cumulus_weighted', 'Cirrocumulus_weighted', 'Nimbus_weighted', 'Clear_weighted'
]

colors = [
    '#FF4B4B','#FF9600','#FFD24B','#AFE182',
    '#82AFE1','#AF96FF','#969696','#000000'
]

df['Date'] = pd.to_datetime(df[['Year', 'Month', 'Day']])
daily_sum = df.groupby('Date')[cloud_types].sum()
daily_sum.index = pd.to_datetime(daily_sum.index)
daily_sum = daily_sum.sort_index()

fig = go.Figure()

# add a bar for each cloud type
for i, cloud_type in enumerate(cloud_types):
    fig.add_trace(go.Bar(
        y=daily_sum.index.strftime('%Y-%m-%d'), x=daily_sum[cloud_type],
        name=cloud_type, marker_color=colors[i],
        orientation='h', showlegend=False
    ))

for cloud_type, color in zip(reversed(cloud_types), reversed(colors)):
    fig.add_trace(go.Scatter(
        x=[None], y=[None], mode='markers',
        marker=dict(size=24, color=color, symbol="square"),
        name=cloud_type.replace('_weighted', ''), showlegend=True, hoverinfo='skip'
    ))

fig.update_layout(
    barmode='stack',
    barnorm='percent',
    yaxis_title='Date',
    xaxis_title='Percentage',
    legend_title='Type',
    font=dict(size=24, family='Arial', weight='bold'),
    yaxis=dict(
        type='category',
        categoryorder='array',
        categoryarray=daily_sum.index.strftime('%Y-%m-%d')
    ),
    width=1500, height=1000,
    margin=dict(l=100, r=30, t=10, b=50),
    legend=dict(
        orientation="h",
        yanchor="bottom",
        x=1, y=1.02,
        xanchor="right"
    )
)

fig.show()
```
